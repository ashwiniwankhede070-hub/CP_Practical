```python
# Grade A - 95 marks and above
# Grade B - 85 marks and above
# Grade C - 75 marks and above
# If marks are below 75, display D
# if elif else format
marks = int(input("Enter your marks: "))
if marks >= 95:
    print("You got grade A")
elif marks >= 85:
    print("You got grade B")
elif marks >= 75:
    print("You got grade C")
else:
    print("You got grade D")
```
